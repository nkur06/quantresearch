```python
ticker_symbol = "HDFCBANK.NS"
```

```python
import yfinance as yf

ticker = yf.Ticker(ticker_symbol)

data = ticker.history(
    period="max",
    interval="1d",
    auto_adjust=False  # optional
)

print(data.head())
print(data.tail())
```

```python
data
```

```python
data.to_csv(f"./1_day_data/{ticker_symbol}.csv")
```
